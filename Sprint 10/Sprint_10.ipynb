### Ejercicio 3 - Columna Hijos con .apply()
_________________________________________________________
- Junta las columnas `Hijos` y `No Hijos` en una sola columna usando `.apply()` y una función definida por ti.
- La nueva columna debe llamarse `Hijos` con valores `Sí` o `No`.

````python
# Junto las columnas `Hijos` y `No Hijos` en una sola columna usando `.apply()`

def combina_columnas(fila):
    if pd.notnull(fila['Fills']) and pd.isnull(fila['No Fills']):
        return 'Si'
    elif pd.notnull(fila['No Fills']) and pd.isnull(fila['Fills']):
        return 'No'
    else:
        return 'Dato raro'


df['Hijos'] = df.apply(combina_columnas, axis=1)
df
````

### Ejercicio 6 - Columna de Edad
_______________________
- Crea una columna `Fecha_Nacimiento` en formato Datetime combinando `Día`, `Mes` y `Año`.
- Define una función que, dada una fecha de nacimiento, calcule la edad actual.
- Aplica la función para crear una nueva columna `Edad` en el DataFrame.

````python
import pandas as pd
from datetime import datetime

df['Fecha_Nacimiento'] = pd.to_datetime(
    df['Año'].astype(str)
    + '-'
    + df['Mes'].astype(str)
    + '-'
    + df['Dia'].astype(str),
    format='%Y-%m-%d',
    errors='coerce',
)
df
````

````python
hoy = datetime.now()
def calcula_edad(fecha_nac):
    edad_dias = (hoy - fecha_nac).days
    return edad_dias // 365 
````

````python
df['Edad'] = df['Fecha_Nacimiento'].apply(calcula_edad)
df = df.drop(columns=['Dia', 'Mes', 'Año'])
df
````

### Ejercicio 2 - Exportar por Grupo Profesional
__________________________
Utilizando un bucle, exporta en 4 archivos (`.xlsx` o `.csv`) los datos de cada Grupo Profesional.
Ejemplo: `datos_GrupoA.xlsx`, `datos_GrupoB.xlsx`, ...

Exporta además un 5º DataFrame con:
- Número de trabajadores por Grupo Profesional
- Sueldo medio
- Edad media

````python
# Utilizando un bucle, exporta en 4 archivos (`.xlsx` o `.csv`) los datos de cada Grupo Profesional.
# Ejemplo: `datos_GrupoA.xlsx`, `datos_GrupoB.xlsx`,
def exporta_grupos(df):
    for grupo, datos_grupo in df.groupby('Grup Professional'):
        nombre_archivo = f'datos_{grupo}.xlsx'
        datos_grupo.to_excel(nombre_archivo, index=False)
````

### Ejercicio 1 - Función de visualización automática
____________________________
Crea una función que tome un DataFrame y genere (y exporte) un gráfico para cada columna:

- **Histograma/Boxplot** si la variable es numérica
- **Barras de valores más frecuentes** si es categórica
- **Barras de años más frecuentes** si está en formato fecha

La función debe funcionar con **cualquier** DataFrame.

Muestra el resultado con un dataset de ejemplo de `seaborn` (ej: `iris`, `penguins` o `titanic`).

````python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
````

````python

df = sns.load_dataset('titanic')
df.dtypes
````

````python
from pandas.api.types import (
    is_bool_dtype,
    is_numeric_dtype,
    is_datetime64_any_dtype
)

def auto_plots(df):
    for col in df.columns:
        serie = df[col].dropna()
        if is_bool_dtype(serie):                            # 1º bolean porque los 1 y o pueden confundirse con números y queremos aplicar gráficos de barras
            plt.figure()
            serie.value_counts().plot(kind='bar')
            plt.title(f'Valores más frecuentes de {col}')
            plt.savefig(f'{col}_bar.png')

        elif is_numeric_dtype(serie):                       # 2º si es numérico
            plt.figure()
            sns.histplot(serie, kde=False)
            plt.title(f'Histograma de {col}')
            plt.savefig(f'{col}_hist.png')

        elif is_datetime64_any_dtype(serie):                # 3º si es fecha
            plt.figure()
            serie.dt.year.value_counts().sort_index().plot(kind='bar')
            plt.title(f'Años más frecuentes en {col}')
            plt.savefig(f'{col}_years.png')

        else:
            plt.figure()                                    # 4º si no es numérico o bool
            serie.value_counts().head(10).plot(kind='bar')
            plt.title(f'Valores más frecuentes de {col}')
            plt.savefig(f'{col}_bar.png')

        plt.close()
````

````python
# Ejercicio 1 - Auto plots
auto_plots(df)
````

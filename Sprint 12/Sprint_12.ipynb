### Ejercicio 1 - Peticiones GET
________________________________________________________________________________________________________________
Consulta la API pública JSONPlaceholder utilizando el método GET para obtener:

- Lista de publicaciones `/posts`.
- Lista de usuarios `/users`.
- Lista de tareas `/todos`.

</div>

```python
# URL base de la API de pruebas
url_base = "https://jsonplaceholder.typicode.com"

# Hago las peticiones GET
respuesta_posts = requests.get(url_base + "/posts")
respuesta_users = requests.get(url_base + "/users")
respuesta_todos = requests.get(url_base + "/todos")

# Convierto las respuestas JSON a objetos de Python
posts = respuesta_posts.json()
users = respuesta_users.json()
todos = respuesta_todos.json()
```

```python
print("POSTS")
print("Cantidad total:", len(posts))
print("Código de estado:", respuesta_posts.status_code)

print("\nUSERS")
print("Cantidad total:", len(users))
print("Código de estado:", respuesta_users.status_code)

print("\nTODOS")
print("Cantidad total:", len(todos))
print("Código de estado:", respuesta_todos.status_code)
```

### Ejercicio 4 - Método POST
________________________________________________________________________________________________________________
Realiza una petición POST para crear una nueva publicación ficticia.

Debe incluir:
- Título.
- Cuerpo del mensaje.
- `userId`.

Muestra la respuesta JSON y el código de estado.

</div>

### Ejercicio 2 - Documentación, endpoints y parámetros
________________________________________________________________________________________________________________
La API de Open Data BCN usa CKAN y devuelve respuestas en formato JSON.

Documentación general:
https://opendata-ajuntament.barcelona.cat/data/es/api/1

Dos endpoints disponibles son:

- `package_search`: permite buscar datasets por texto.
- `package_show`: permite consultar el detalle de un dataset concreto.
- También usaré `datastore_search`, que permite recuperar registros de un recurso.

Parámetros interesantes:

- `q`: texto de búsqueda en `package_search`.
- `rows`: número de resultados que queremos recuperar.
- `id`: identificador del dataset en `package_show`.
- `resource_id`: identificador del recurso en `datastore_search`.
- `limit`: número de registros que queremos recuperar.

</div>

<div style="background-color: #c5c8cb; padding: 10px; border-radius: 5px;">

### Ejercicio 3 - Petición GET sencilla
________________________________________________________________________________________________________________
Hago una búsqueda sencilla del dataset IRIS y muestro algunos campos de la respuesta JSON.

</div>

```python
url_bcn = "https://opendata-ajuntament.barcelona.cat/data/api/3/action"

parametros_iris = {
    "q": "iris",
    "rows": 3
}

respuesta_iris_search = requests.get(url_bcn + "/package_search", params=parametros_iris)
datos_iris_search = respuesta_iris_search.json()

print("Código de estado:", respuesta_iris_search.status_code)
print("Búsqueda correcta:", datos_iris_search["success"])
print("Resultados encontrados:", datos_iris_search["result"]["count"])

primer_resultado = datos_iris_search["result"]["results"][0]
print("Nombre del dataset:", primer_resultado["name"])
print("Título:", primer_resultado["title"])
print("ID:", primer_resultado["id"])
```

<div style="background-color: #c5c8cb; padding: 10px; border-radius: 5px;">

### Ejercicio 4 - Convertir datos a DataFrame
________________________________________________________________________________________________________________
Convierto los resultados de la búsqueda en un DataFrame de pandas.

</div>

```python
resultados_iris = datos_iris_search["result"]["results"]

df_busqueda_iris = pd.DataFrame(resultados_iris)

df_busqueda_iris[["name", "title", "id"]].head()
```

### Ejercicio 2 - Ver detalle con package_show
________________________________________________________________________________________________________________
Selecciono el dataset `iris` y reviso sus recursos disponibles con `package_show`.

Después escojo un recurso CSV con `datastore_active = True` para poder consultarlo con `datastore_search`.

Uso el recurso de 2024 porque es un año cerrado y tiene datastore activo.

</div>

```python
id_dataset = "iris"

respuesta_detalle = requests.get(url_bcn + "/package_show", params={"id": id_dataset})
datos_detalle = respuesta_detalle.json()

print("Código de estado:", respuesta_detalle.status_code)
print("Dataset seleccionado:", datos_detalle["result"]["title"])
print("\nRecursos disponibles:")

for recurso in datos_detalle["result"]["resources"]:
    print("Nombre:", recurso.get("name"))
    print("Formato:", recurso.get("format"))
    print("Datastore activo:", recurso.get("datastore_active"))
    print("Resource id:", recurso.get("id"))
    print("---")
```

```python
# Elijo el CSV de 2024 porque tiene datastore activo
recurso_elegido = None

for recurso in datos_detalle["result"]["resources"]:
    nombre = recurso.get("name", "")
    formato = recurso.get("format", "").upper()
    datastore_activo = recurso.get("datastore_active", False)
    
    if "2024" in nombre and formato == "CSV" and datastore_activo == True:
        recurso_elegido = recurso
        break

print("Recurso elegido:", recurso_elegido["name"])
print("Formato:", recurso_elegido["format"])
print("Resource id:", recurso_elegido["id"])
```

<div style="background-color: #c5c8cb; padding: 10px; border-radius: 5px;">

### Ejercicio 3 - Consulta con datastore_search
________________________________________________________________________________________________________________
Con el `resource_id` del recurso seleccionado, hago una consulta con `datastore_search` para recuperar al menos 100 registros.

</div>

```python
resource_id = recurso_elegido["id"]

parametros_datastore = {
    "resource_id": resource_id,
    "limit": 100
}

respuesta_datastore = requests.get(url_bcn + "/datastore_search", params=parametros_datastore)
datos_datastore = respuesta_datastore.json()

registros = datos_datastore["result"]["records"]

print("Código de estado:", respuesta_datastore.status_code)
print("Consulta correcta:", datos_datastore["success"])
print("Total de registros del recurso:", datos_datastore["result"]["total"])
print("Registros recuperados:", len(registros))
```

<div style="background-color: #c5c8cb; padding: 10px; border-radius: 5px;">

### Ejercicio 4 - Convertir resultados a DataFrame
________________________________________________________________________________________________________________
Convierto los registros obtenidos desde la API en un DataFrame de pandas.

</div>

```python
df_iris_incidencias = pd.DataFrame(registros)
df_iris_incidencias.head()
```

<div style="background-color: #c5c8cb; padding: 10px; border-radius: 5px;">

### Ejercicio 5 - Guardar DataFrame en CSV
________________________________________________________________________________________________________________
Guardo el DataFrame en un archivo `.csv`.

```python
nombre_archivo = "iris_incidencias_bcn_2024.csv"

df_iris_incidencias.to_csv(nombre_archivo, index=False)

print("Archivo guardado:", nombre_archivo)
print("Filas guardadas:", len(df_iris_incidencias))
print("Columnas guardadas:", len(df_iris_incidencias.columns))
```
